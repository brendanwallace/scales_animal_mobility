```python
import os
import glob
import pandas as pd
```

```python
START = 60

csvs = glob.glob("../data/movebank/events/*.csv")
for i in range(START, len(csvs)):
    old_name = csvs[i]
    
    print(old_name)
    df = pd.read_csv(old_name)
    # There should only be one study id.
    if len(df) < 1 or "study_id" not in df.columns:
        continue
    assert(len(df["study_id"].unique()) == 1)
    study_id = df["study_id"].unique()[0]
    new_name = f"../data/movebank/events/{study_id}.csv"
    os.rename(old_name, new_name)
    print(new_name)
```
