```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import glob

# Change these to wherever your files are.
DATA = "../results/S1_stop_radii/stops_at_different_radii/*.csv"
METADATA = "../data/species_metadata.csv"

STOP_RADII = [1, 2, 3, 5, 10, 15, 20, 30, 40, 50, 60, 80, 100, 150, 200]

sns.set_context("talk")
pd.options.display.max_columns = None
```

```python
full_df = pd.concat((pd.read_csv(f) for f in glob.glob(DATA)), ignore_index=True)

species_metadata = pd.read_csv(METADATA)
full_df = pd.merge(full_df, species_metadata, on=["species"])
full_df["movement"] = pd.Categorical(full_df["movement"], ['WALK', 'FLY', 'SWIM'])

df = full_df

df = df[df["days_spanned"] >= 60]
```

```python
# Computes 90th% and 70th% radii.
best_radii = []
for species, sdf in df.groupby("species"):
    best_radius, best_stops = None, -1
    _dict = {
        "species": species,
        "best_radius": -1,
    }
    # first find the best.
    for stop_radius, rdf in sdf.groupby("stop_radius"):
        if rdf.num_stops.sum() > best_stops:
            best_stops = rdf.num_stops.sum()
            best_radius = stop_radius
    _dict["best_radius"] = best_radius
    # find 70th & 90th percents
    for p in [0.7, 0.9]:
        for stop_radius, rdf in sdf.groupby("stop_radius"):
            if rdf.num_stops.sum() >= best_stops * p:
                # find the first time we're at the proportion, and then exit.
                _dict[f"{p}_radius"] = stop_radius
                break

    best_radii.append(_dict)

best_radii = pd.DataFrame(best_radii)

best_radii.to_csv("../results/best_radii_by_species.csv")
```
