Near the equator, latitude and longitude are close to square with each degree about 111 km.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
NUM_STEPS = 1000

def random_walk(num_steps):
    x = np.zeros((num_steps+1, 2))
    for i in range(num_steps):
        
        x[i+1] = x[i] + (np.random.rand(2) - 0.5)
    return x
```

```python
for r in range(10):
    x = random_walk(NUM_STEPS)
    plt.plot(x[:,0], x[:, 1])
```
